```python
sales_by_category = cleaned_data.groupby("Category")["Sales"].sum()
```

```python
sns.barplot(
    x=sales_by_category.index,
    y=sales_by_category.values
)

plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
profit_by_category = cleaned_data.groupby("Category")["Profit"].sum()

print(profit_by_category)
```

```python
sns.barplot(
    x=profit_by_category.index,
    y=profit_by_category.values
)

plt.title("Total Profit by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.show()
```

```python
profit_by_discount = cleaned_data.groupby("Discount")["Profit"].mean()

print(profit_by_discount)
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    x=profit_by_discount.index,
    y=profit_by_discount.values,
    marker="o"
)

plt.title("Impact of Discounts on Average Profit")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.grid(True)

plt.show()
```

```python
cleaned_data["Year-Month"] = cleaned_data["Order Date"].dt.to_period("M")
```

```python
monthly_sales = cleaned_data.groupby("Year-Month")["Sales"].sum()

print(monthly_sales.head(10))
```

```python
plt.figure(figsize=(14, 6))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title("Monthly Sales Trend (2015–2018)")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
profit_by_region = cleaned_data.groupby("Region")["Profit"].sum()

print(profit_by_region)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=profit_by_region.index,
    y=profit_by_region.values
)

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.show()
```

```python
profit_by_subcategory = cleaned_data.groupby("Sub-Category")["Profit"].sum()

print(profit_by_subcategory.sort_values(ascending=False))
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    x=profit_by_subcategory.sort_values().values,
    y=profit_by_subcategory.sort_values().index
)

plt.title("Profit by Product Sub-Category")
plt.xlabel("Total Profit")
plt.ylabel("Sub-Category")

plt.axvline(x=0, color="black", linewidth=1)

plt.tight_layout()
plt.show()
```

```python
category_analysis = cleaned_data.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})

category_analysis["Profit Margin (%)"] = (
    category_analysis["Profit"] / category_analysis["Sales"]
) * 100

print(category_analysis)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    category_analysis.index,
    category_analysis["Profit Margin (%)"]
)

plt.title("Profit Margin by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Profit Margin (%)")

plt.grid(axis="y")
plt.show()
```

```python
correlation = cleaned_data["Sales"].corr(cleaned_data["Profit"])

print("Correlation between Sales and Profit:", correlation)
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=cleaned_data,
    x="Sales",
    y="Profit"
)

plt.title("Relationship Between Sales and Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True)
plt.show()
```

```python
top_products = (
    cleaned_data.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Most Profitable Products")
plt.xlabel("Total Profit")
plt.ylabel("Product Name")
plt.grid(axis="x")

plt.tight_layout()
plt.show()
```

```python
least_products = (
    cleaned_data.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=True)
    .head(10)
)

print(least_products)
```

```python
plt.figure(figsize=(10, 6))

least_products.sort_values().plot(kind="barh")

plt.title("Top 10 Least Profitable Products")
plt.xlabel("Total Profit")
plt.ylabel("Product Name")
plt.grid(axis="x")

plt.tight_layout()
plt.show()
```

# Final Findings and Business Recommendations

## Key Findings

1. Technology generates the highest sales and total profit.
2. Furniture has the lowest profit margin at approximately 2.49%.
3. The West region generates the highest total profit.
4. Higher discount levels generally show lower average profit.
5. Sales and profit have a moderate positive correlation of 0.479.
6. Canon imageCLASS 2200 Advanced Copier is the most profitable product.
7. Cubify CubeX 3D Printer Double Head Print has the highest total loss.
8. Tables and Bookcases have negative total profit.
9. Monthly sales fluctuate, with some strong peaks towards the end of 2018.

## Business Recommendations

1. Focus on maintaining sales and profitability in the Technology category.
2. Review Furniture pricing and operational costs to improve its profit margin.
3. Investigate products with consistently negative profits.
4. Review discount strategies to avoid unnecessary reductions in profitability.
5. Study successful products and regions to identify opportunities for growth.

## Conclusion

The analysis of the Superstore dataset reveals differences in sales,
profitability, product performance, and regional performance.
The findings can help the business improve pricing strategies,
manage discounts, identify loss-making products, and make
better business decisions.

# Final Findings and Conclusion

## Key Findings

1. **Category Performance:** Technology generated the highest total profit (approximately 145,455), followed by Office Supplies (122,491) and Furniture (18,451).

2. **Profit Margins:** Technology had the highest profit margin at approximately 17.40%, while Furniture had the lowest at approximately 2.49%.

3. **Regional Performance:** The West region generated the highest total profit (approximately 108,418), followed by East, South, and Central.

4. **Discount Impact:** Higher discount levels generally showed lower average profits. Some discount levels resulted in negative average profits.

5. **Product Performance:** Canon imageCLASS 2200 Advanced Copier was the most profitable product, while Cubify CubeX 3D Printer Double Head Print recorded the highest total loss.

6. **Sub-Category Performance:** Copiers generated the highest profit, whereas Tables recorded the largest loss.

7. **Sales and Profit Relationship:** Sales and Profit showed a moderate positive correlation of approximately 0.479.

## Conclusion

The analysis of the Superstore dataset revealed differences in sales, profitability, product performance, and regional performance.

The findings suggest that the business could improve profitability by reviewing discount strategies, identifying loss-making products, improving Furniture category performance, and focusing on profitable products and regions.

Data visualization helped identify important patterns and trends, supporting better business decision-making.
